## 1. Basic Evaluation (Relevance & Completeness)
Evaluate whether a response addresses the prompt and covers all necessary sub-questions.

```python
prompt = "Explain what an API is and provide two common HTTP methods."
response = (
    "An API allows different software applications to communicate. "
    "Two common HTTP methods are GET (retrieve) and POST (submit)."
)

result = evaluate(
    prompt=prompt,
    response=response,
)

print(f"Relevance:    {result.relevance * 100:.1f}%")
print(f"Completeness: {result.completeness * 100:.1f}%")
print(f"Overall:      {result.overall * 100:.1f}%")
```

## 2. RAG Evaluation (Faithfulness & Hallucination Detection)
Pass retrieved context to verify factual consistency.

```python
rag_context = (
    "The James Webb Space Telescope launched on Dec 25, 2021 from French Guiana."
)
faithful_response = (
    "The James Webb Space Telescope launched on Christmas Day 2021 from Guiana."
)

rag_result = evaluate(
    prompt="When and where was the James Webb Space Telescope launched?",
    response=faithful_response,
    context=rag_context,
)

print(f"Faithfulness:  {rag_result.faithfulness * 100:.1f}%")
print(f"Hallucination: {rag_result.hallucination * 100:.1f}%")
print(f"Overall Score: {rag_result.overall * 100:.1f}%")
```

## 3. Custom Metric Weighting
Configure custom weights prioritizing faithfulness for compliance or accuracy-critical pipelines.

```python
compliance_config = EvaluatorConfig(
    relevance_weight=0.20,
    completeness_weight=0.20,
    faithfulness_weight=0.60,
)

compliance_result = evaluate(
    prompt="When was JWST launched?",
    response=faithful_response,
    context=rag_context,
    config=compliance_config,
)

print("Serialized Result JSON:")
print(compliance_result.to_json(indent=2))
```
